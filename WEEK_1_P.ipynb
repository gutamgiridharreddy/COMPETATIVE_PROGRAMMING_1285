```python
import sys

data = sys.stdin.read().strip().split()
if not data:
    data = """1
8 3
101 101 102 102 102 105 110 110
101
102
999""".split()

i = 0

def first(a,x):
    l,r,ans=0,len(a)-1,-1
    while l<=r:
        m=(l+r)//2
        if a[m]>=x:
            if a[m]==x: ans=m
            r=m-1
        else:
            l=m+1
    return ans

def last(a,x):
    l,r,ans=0,len(a)-1,-1
    while l<=r:
        m=(l+r)//2
        if a[m]<=x:
            if a[m]==x: ans=m
            l=m+1
        else:
            r=m-1
    return ans

t = int(data[i]); i += 1

for _ in range(t):
    n = int(data[i]); i += 1
    q = int(data[i]); i += 1

    a = list(map(int, data[i:i+n]))
    i += n

    for _ in range(q):
        x = int(data[i]); i += 1
        f,l = first(a,x), last(a,x)
        print(0 if f==-1 else l-f+1)
```

```python
def merge_sort(arr):
    if len(arr) <= 1: return arr
    mid = len(arr)//2
    left = merge_sort(arr[:mid]); right = merge_sort(arr[mid:])
    i=j=0; res=[]
    while i<len(left) and j<len(right):
        if left[i] <= right[j]: res.append(left[i]); i+=1
        else: res.append(right[j]); j+=1
    return res + left[i:] + right[j:]

data = """1
7
4 1 6 2 5 3 2"""

it = iter(data.split())
t = int(next(it))

for _ in range(t):
    n = int(next(it))
    arr = [int(next(it)) for _ in range(n)]
    print(*merge_sort(arr))
```
